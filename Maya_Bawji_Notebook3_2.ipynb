Question 1: Identify all the outliers in the above data.  Justify your answers using objective measures.

```python
Q1 = grads['Avg Hours Studies per Week'].quantile(0.25)
Q3 = grads['Avg Hours Studies per Week'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print(Q1)
print(Q3)
print(IQR)
print(lower_limit)
print(upper_limit)
```

```python
Q1 = grads['GPA'].quantile(0.25)
Q3 = grads['GPA'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print(Q1)
print(Q3)
print(IQR)
print(lower_limit)
print(upper_limit)
```

```python
Q1 = grads['Sense of Humour (0-5)'].quantile(0.25)
Q3 = grads['Sense of Humour (0-5)'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print(Q1)
print(Q3)
print(IQR)
print(lower_limit)
print(upper_limit)
```

### Answer

I used the IQR method to identify the outliers.

- For Avg Hours Studies per Week, the limits are -25 and 95. The value 100 is above the upper limit, so it is an outlier.
- For GPA, the limits are 1.3125 and 5.0125. There are no outliers.
- For Sense of Humour, the limits are -3.275 and 7.725. There are no outliers.

Therefore, the only outlier is 100 in Avg Hours Studies per Week.

Question 2: There are various data that are missing.  Fill-in the missing data or delete the rows/columns that you think you should delete.  Justify your answer

```python
grads['GPA'] = grads['GPA'].fillna(grads['GPA'].median())
grads['Sense of Humour (0-5)'] = grads['Sense of Humour (0-5)'].fillna(grads['Sense of Humour (0-5)'].median())

grads
```

### Justification

I filled the missing numerical values using the median. I did not delete the rows because the dataset is small and deleting rows would remove useful information. The median is less affected by extreme values, so it is appropriate for this dataset.

Exercise II. Run the cell below to create a new dataframe called `df_miss`.  Its first column will contain some missing values.

```python
import pandas as pd
import numpy as np
import random

nrows = 10
ncols = 5

# set a seed for random number generation
np.random.seed(314)
# create an array filled with random data
data = np.array(np.random.rand(nrows, ncols))
# put the data to a pandas dataframe
df_miss = pd.DataFrame(data)
# rename the columns
df_miss.columns = ['col_'+str(ii) for ii in range(ncols)]

# randomly set some values to missing
ix0 = np.random.randint(nrows, size=3)
ix1 = np.random.randint(nrows, size=3)

df_miss['col_0'][ix0] = np.nan
df_miss['col_1'][ix1] = np.nan

print(df_miss)
```

Impute the missing values (NaN) in `col_0` (but not `col_1`) with the median.  Store the values in the dataframe by using the parameter `inplace`.  Print the dataframe.

```python
df_miss['col_0'].fillna(df_miss['col_0'].median(), inplace=True)

print(df_miss)
```

Impute the missing values in `col_1` with value 0.  Store the values in the dataframe by using the parameter `inplace`.  Print the dataframe.

```python
df_miss['col_1'].fillna(0, inplace=True)

print(df_miss)
```
